```python
question = "What is covered in CS 450?"

ground_truth = ["CS 450"]*5

retrieved = ["CS 450"]*5
```

```python
def evaluate_retrieval(ground_truth, retrieved):
    """
    Evaluate the retrieval system based on precision, recall, and F1 score.
    
    Args:
    """
    true_positives = 0
    for i in range(len(ground_truth)):
        if ground_truth[i] == retrieved[i]:
            true_positives += 1
    
    precision = true_positives / len(retrieved)
    recall = true_positives / len(ground_truth)
    f1 = 2 * (precision * recall) / (precision + recall)
    return precision, recall, f1

precision, recall, f1 = evaluate_retrieval(ground_truth, retrieved)
f1

```
